```python
def change_making(coint_set, value):
    # coin_set is a list of coin values sorted in decreasing order
    # value is the amount to give back
    # return the list of coins to give back

    ########################
    #
    # Your code here !
    #
    ########################
    if not coint_set:
        return []


    res = []

    while value > 0:
        for i in range(len(coint_set)):
            if coint_set[i] <= value:
                res.append(coint_set[i])
                value -= coint_set[i]
                break

    return res

test_change_making(change_making);



def gain_robber(weights, price, max_weight):
    res = []
    n = len(weights)

    if not weights or not price or max_weight <= 0:
        return res
    
    total_store = []

    for i in range(n):
        total_store.append([price[i]/weights[i], weights[i], price[i], i])

    total_store.sort(reverse=True)

    remaining_weight = max_weight
    fractions = [0]*n
    total_value = 0

    for d, w, p, i in total_store:
        if remaining_weight <= 0:
            break

        if remaining_weight >= w:
            fractions[i] = 1
            total_value += p
            remaining_weight -= w
        else:
            frac = remaining_weight / w
            fractions[i] = frac
            total_value += p * frac
            remaining_weight = 0

    return fractions, total_value


print(gain_robber([10, 20, 30], [60, 100, 120], 50))
print(gain_robber([10, 20, 30], [100, 180, 120], 50))
print(gain_robber([30, 20, 10], [60, 100, 120], 50))
```

### Activity selection problem

**Implement a greedy solution to the activity selection problem.**

The unit test must says that your function seems correct.

```python
def activity_selection(activities):
    """
    Input: list of activities as pairs (start, finish)
    Output: maximal list of non overlapping activities

    Example:
    >>activity_selection([(0, 3), (6, 7), (2, 6), (1, 2), (8, 9), (3, 5), (5, 9), (3, 6)])
    [(1, 2), (3, 5), (6, 7), (8, 9)]
    """
    ##################
    # YOUR CODE HERE
    ##################
    n = len(activities)
    if n <= 1:
        return activities

    activities.sort(key=lambda x: x[1])  # Sort by finish time

    result = []
    result.append(activities[0])

    for i in range(1, n):
        if activities[i][0] >= result[-1][1]:
            result.append(activities[i])

    return result


test_activity_selection(activity_selection);


def kemit_jump(n, delta):

    current_index = 0
    result = []

    while current_index != n:
        max_jump_possible = min(delta[current_index], n)
        current_index = max_jump_possible
        result.append(current_index)

    return result, len(result)


def kermit_jump_greedy(n, delta):
    jump = 0
    currentEnd = 0
    nextEnd = 0


    for i in range(0, n - 1):
        nextEnd = max(nextEnd, delta[i])

        if i == currentEnd:
            jump += 1
            currentEnd = nextEnd

    return jump


print(kemit_jump(6, [4, 1, 1, 6, 5, 6, 6]))
print(kermit_jump_greedy(6, [4, 1, 1, 6, 5, 6, 6]))
```

### Huffman coding

The following class represents a Node of a binary tree for Huffman coding.

```python
class Node:
    """
    Node of a binary tree. A node has 4 attributes:
    - weight (mandatory)
    - symbol (optional)
    - left_child (optional)
    - right_child (optional)
    """

    def __init__(self, weight, symbol='', left_child=None, right_child=None):
        self.weight = weight
        self.symbol = symbol
        self.left_child = left_child
        self.right_child = right_child

    def __repr__(self):
        """
        String representation of the node: (weight, symbol, left_child, right_child)
        """
        def str_rec(node):
            if node is None:
                return ""
            else:
                return str(node)
        return "(" + str(self.weight) + ", " + str(self.symbol) + ", " + str_rec(self.left_child) + ", " + str_rec(self.right_child) +")"
    
    def __eq__(self, other):
        """
        Comparison function == between two nodes.
        """
        if not(isinstance (other, Node)):
            return False
        return self.weight == other.weight and self.symbol == other.symbol and self.left_child == other.left_child and self.right_child == other.right_child

    def __lt__(self, other):
        """
        Comparison function < between two nodes.
        """

        return self.weight < other.weight
    
def print_demo_tree():
    """
    Construct and print a demo tree
    """
    n1 = Node(5, "a")
    n2 = Node(6, "b")
    n3 = Node(10, "c")
    n4 = Node(11, left_child=n1, right_child=n2)
    root = Node(21, left_child=n3, right_child=n4)
    print("Displays a small tree:")
    print(root)

print_demo_tree()
```
