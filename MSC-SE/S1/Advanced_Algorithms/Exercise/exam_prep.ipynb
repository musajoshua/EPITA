```python
def find_peak_elements(array):
    peak_elem = []
    array_len = len(array)
	
    if array == []:
	    return peak_elem

    if array_len == 1:
	    peak_elem.append(array_len -1)
	    return peak_elem
	

    if array[0] > array[1]:
        peak_elem.append(0)

    if array[-1] > array[-2]:
        peak_elem.append(len(array) -1)

    for i in range(1, array_len -1 ):

        if array[i] > array[i + 1] and array[i] > array[i - 1]:
            peak_elem.append(i)

    return peak_elem

def find_peak_element(array):
	array_len = len(array)
	
	if array == []:
	    return -1

	if array_len == 1:
	    return 0
	

	if array[0] > array[1]:
		return 0

	if array[-1] > array[-2]:
		return array_len -1

	for i in range(1, array_len -1 ):
		if array[i] > array[i + 1] and array[i] > array[i - 1]:
			return i

def peak_element_iterative(array): # O(n)
	if array == []:
		return -1
	if len(array)==1:
		return 0
	if array[0]>array[1]:
		return 0
	if array[len(array)-1]>array[len(array)-2]:
		return len(array)-1
	for i in range(1,len(array)-1):
		# if I have a peak 
		if array[i]>array[i+1] and array[i]>array[i-1]:
			return i

print(find_peak_element([0, 2, 3, 1, 5, 4, 6, 9]))
print(find_peak_elements([0, 2, 3, 1, 5, 4, 6, 9]))
print(peak_element_iterative([0, 2, 3, 1, 5, 4, 6]))
```
