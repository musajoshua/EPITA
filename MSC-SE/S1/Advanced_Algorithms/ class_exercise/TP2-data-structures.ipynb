### Intersection of two lists

**Implement `inter` in the following function.**

You can use the `sort` function, you cannot use a `dictionary` or a `set`.

The unit test must says that your function seems correct.

```python
def inter_slow(a, b):
    ########################
    #
    # Your code here !    
    #
    ########################
    # a.sort()
    # b.sort()

    intersection = []

    for i in range(len(a)):
        for j in range(len(b)):
            if a[i] == b[j]:
                intersection.append(a[i])
                break
    
    unique_intersection = unique(intersection)

    return unique_intersection

def inter(a, b): # O(nLog(n) + mLog(m) + n + m)
    ########################
    #
    # Your code here !    
    #
    ########################
    intersection = []
    a.sort()
    b.sort()

    index_a = 0 # nLog(n)
    index_b = 0 # mLog(m)

    while index_a < len(a) and index_b < len(b): # O(n + m)
        if a[index_a] == b[index_b]:
            if a[index_a] not in intersection:
                intersection.append(a[index_a])
            index_a += 1
            index_b += 1
        elif a[index_a] < b[index_b]:
            index_a += 1
        else:
            index_b += 1

    return intersection

test_inter(inter);
```

### Reverse Polish Notation

**Implement `reverse_polish_evaluation` in the following function.**

The input is a string containing integers and symbols + and * encoding a mathematical expression in reverse Polish notation and the function returns the value of the expression.
 
The unit test must says that your function seems correct.

```python
def reverse_polish_evaluation(s):
    ########################
    #
    # Your code here !    
    #
    ########################
    tokens = s.split()
    print(tokens)
    stack = []

    for token in tokens:
        if token == '+':
            stack.append(stack.pop() + stack.pop())
        elif token == '*':
            stack.append(stack.pop() * stack.pop())
        elif token == '-':
            stack.append(stack.pop() - stack.pop())
        elif token == '/':
            stack.append(stack.pop() / stack.pop())
        else:
            stack.append(int(token))
    print(stack)
    return stack[0]
            
            

test_reverse_polish_evaluation(reverse_polish_evaluation);   
```

* `s.append(e)`: push a new element on the top of the stack
* `s.pop()`: remove and return the element at the top of the stack

The unit test must says that your function seems correct.

```python
def is_balanced(s):
    ########################
    #
    # Your code here !    
    #
    ########################
    
    tokens = s.split()
    stack = []

    for token in tokens:
        if token == '(':
            stack.append(token)
        elif token == '[':
            stack.append(token)
        elif token == ')':
            poppedElement = stack.pop()
            if poppedElement != '(':
                stack.append(poppedElement)
                break
        elif token == ']':
            poppedElement = stack.pop()
            if poppedElement != '[':
                stack.append(poppedElement)
                break
    
    print(stack)

    if len(stack) == 0:
        return True
    else:
        return False

test_is_balanced(is_balanced);
```
