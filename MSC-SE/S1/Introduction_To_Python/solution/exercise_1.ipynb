
Exercise 3-2
1.	Ask the user for two numbers 
2.	Assign these numbers into the variables a and b.
3.	Convert the numbers into integers.
4.	The goal is to swap the values in such a way that: 
a.	a gets the value of b. 
b.	b gets the value of a.
5.	Now, reset a and b to their original values. This time, swap the values of a and b using a different method.


```python
a, b = input("Enter first number"), input("Enter second number")

num_1, num_2 = int(a), int(b)
print("num_1 is", num_1)
print("num_2 is", num_2)

c = num_1
num_1 = num_2
num_2 = c

print("num_1 is", num_1)
print("num_2 is", num_2)


num_1, num_2 = num_2, num_1
 # or
# num_1 = num_1 + num_2
# num_2 = num_1 - num_2
# num_1 = num_1 - num_2

print("num_1 is", num_1)
print("num_2 is", num_2)
```
