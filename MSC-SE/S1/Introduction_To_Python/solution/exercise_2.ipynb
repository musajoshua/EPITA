Exercise 2-1
    A store offers the following discounts:
    Purchase over $50 but less than $100: 10% discount.
    Purchase of $100 or more: 20% discount.
    Ask the user for the amount of their purchase, and calculate the total amount after applying the discount.

```python
amount = int(input("Enter your amount of purchase"))

if amount > 50 and amount < 100:
    amount *= 0.9
elif amount >= 100:
    amount *= 0.8

print("Final amount to pay", amount)
```

    Ask the user for the lengths of three sides of a triangle. 
    Determine if those sides can form a valid triangle. (Hint: In a valid triangle, the sum of any two sides should be greater than the third side.)

```python
side_1 = int(input("Enter the length of first side of the triangle"))
side_2 = int(input("Enter the length of second side of the triangle"))
side_3 = int(input("Enter the length of third side of the triangle"))

is_valid = False

if side_1 + side_2 > side_3:
    is_valid = True
elif side_1 + side_3 > side_2:
    is_valid = True
elif side_2 + side_3 > side_1:
    is_valid = True

if is_valid:
    print("Not a valid triangle")
else:
    print("Valid Triangle")
```


Exercise 2-4
    Ask the user for their weight (in kilograms) and height (in meters). 
    Calculate their Body Mass Index (BMI) using the formula:
BMI = Weight (kg) / Height(m)²
    Then categorize their BMI into:
    Underweight: <18.5
    Normal weight: 18.5–24.9
    Overweight: 25–29.9
    Obesity: ≥30
    Print out their BMI and its category.

```python
weight = float(input("Enter weight in kg"))
height = float(input("Enter height in meters"))

bmi = round(weight / (height ** 2), 1)

category = ''

if bmi < 18.5:
    category = 'Underweight'
elif bmi >= 18.5 and bmi <= 24.9:
    category = 'Normal Weight'
elif bmi >= 25.0 and bmi <= 29.9:
    category = 'Overweight'
elif bmi >= 30.0:
    category = "Obesity"

print("BMI is", bmi, "and category is", category)
```
